```python
# Laptop

class Laptop:
    def __init__(self, brand, ram, storage, price):
        self.brand = brand
        self.ram = ram
        self.storage = storage
        self.price = price

    def display_laptop(self):
        print("\n--- Laptop Details ---")
        print("Brand:", self.brand)
        print("RAM:", self.ram)
        print("Storage:", self.storage)
        print("Price:", self.price)


# Store multiple laptops data
laptops = []

# User input for multiple laptops
number = int(input("How many laptops do you want to enter? "))

for i in range(number):
    print(f"\nEnter Laptop {i + 1}")

    brand = input("Enter laptop brand: ")
    ram = input("Enter laptop RAM: ")
    storage = input("Enter laptop storage: ")
    price = int(input("Enter laptop price: "))

    # Create object
    laptop = Laptop(
        brand,
        ram,
        storage,
        price
    )

    # Add object to list
    laptops.append(laptop)


# Display all laptops
print("\n===== ALL LAPTOPS =====")

for laptop in laptops:
    laptop.display_laptop()
```
