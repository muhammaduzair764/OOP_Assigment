```python
# Shopping Product

class Product:
    def __init__(self, product_name, category, price, stock_quantity):
        self.product_name = product_name
        self.category = category
        self.price = price
        self.stock_quantity = stock_quantity

    def display_product(self):
        print("\n--- Product Details ---")
        print("Product Name:", self.product_name)
        print("Category:", self.category)
        print("Price:", self.price)
        print("Stock Quantity:", self.stock_quantity)


# Store multiple products data
products = []

# User input for multiple products
number = int(input("How many products do you want to enter? "))

for i in range(number):
    print(f"\nEnter Product {i + 1}")

    product_name = input("Enter product name: ")
    category = input("Enter product category: ")
    price = int(input("Enter product price: "))
    stock_quantity = int(input("Enter stock quantity: "))

    # Create object
    product = Product(
        product_name,
        category,
        price,
        stock_quantity
    )

    # Add object to list
    products.append(product)


# Display all products
print("\n===== ALL SHOPPING PRODUCTS =====")

for product in products:
    product.display_product()
```
