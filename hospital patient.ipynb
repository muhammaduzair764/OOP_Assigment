```python
# Hospital Patient

class Hospital:
    def __init__(self, patient_name, patient_age, patient_gender, patient_disease):
        self.patient_name = patient_name
        self.patient_age = patient_age
        self.patient_gender = patient_gender
        self.patient_disease = patient_disease

    def display_patient(self):
        print("\n--- Patient Details ---")
        print("Patient Name:", self.patient_name)
        print("Patient Age:", self.patient_age)
        print("Patient Gender:", self.patient_gender)
        print("Patient Disease:", self.patient_disease)


# Store multiple patients data
patients = []

# User input for multiple patients
number = int(input("How many patients do you want to enter? "))

for i in range(number):
    print(f"\nEnter Patient {i + 1}")

    patient_name = input("Enter patient name: ")
    patient_age = int(input("Enter patient age: "))
    patient_gender = input("Enter patient gender: ")
    patient_disease = input("Enter patient disease: ")

    # Create object
    patient = Hospital(
        patient_name,
        patient_age,
        patient_gender,
        patient_disease
    )

    # Add object to list
    patients.append(patient)


# Display all patients
print("\n===== ALL HOSPITAL PATIENTS =====")

for patient in patients:
    patient.display_patient()
```
