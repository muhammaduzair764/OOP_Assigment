```python
# Mobile Phone Details
class Mobile_Phone:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def display_info(self):
        print("\nMobile Phone Details:")
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Price:", self.price)


# Ask how many mobile phones
number = int(input("How many Mobile Phones do you want to enter? "))

# List to store objects
mobile_phones = []

# Input multiple mobile phones
for i in range(number):

    print("\nEnter Mobile Phone", i + 1, "Details")

    brand = input("Enter Brand Name: ")
    model = input("Enter Model: ")
    price = input("Enter Price: ")

    # Create object
    phone = Mobile_Phone(brand, model, price)

    # Add object to list
    mobile_phones.append(phone)


# Display all mobile phones
print("\n========== ALL MOBILE PHONES ==========")

for phone in mobile_phones:
    phone.display_info()bd
```
