```python
# Restaurant Order Class:

class RestaurantOrder:
    def __init__(self, customer_name, food_item, quantity):
        self.customer_name = customer_name
        self.food_item = food_item
        self.quantity = quantity

    def display_order(self):
        print("\n--- Order Details ---")
        print("Customer Name:", self.customer_name)
        print("Food Item:", self.food_item)
        print("Quantity:", self.quantity)


# Store multiple orders:
orders = []

# User input for multiple customers
number = int(input("How many orders do you want to enter? "))

for i in range(number):
    print(f"\nEnter Order {i + 1}")

    customer_name = input("Enter customer name: ")
    food_item = input("Enter food item: ")
    quantity = int(input("Enter quantity: "))

    # Create object
    order = RestaurantOrder(customer_name, food_item, quantity)

    # Add object to list
    orders.append(order)


# Display all orders
print("\n===== ALL RESTAURANT ORDERS =====")

for order in orders:
    order.display_order()
```
