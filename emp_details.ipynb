```python
class Book:
    def __init__(self, title, author_name, price):
        self.title = title
        self.author_name = author_name
        self.price = price

    def display_info(self):
        print("\nBook Details:")
        print("Book Title:", self.title)
        print("Book Author Name:", self.author_name)
        print("Book Price:", self.price)


# Ask how many books
number = int(input("How many books do you want to enter? "))

books = []

# Get data for multiple books
for i in range(number):


    title = input("Enter book title: ")
    author_name = input("Enter book author name: ")
    price = int(input("Enter book price: "))

    book = Book(title, author_name, price)
    books.append(book)


# Display all books
print("\n========== ALL BOOKS ==========")

for book in books:
    book.display_info()
```

```python
# Employee Details

class Employee:
    def __init__(self, name, id, department, salary):
        self.name = name
        self.id = id
        self.department = department
        self.salary = salary

    def display_info(self):
        print("\nEmployee Details:")
        print("Employee Name:", self.name)
        print("Employee ID:", self.id)
        print("Employee Department:", self.department)
        print("Employee Salary:", self.salary)


# Ask how many employees
number = int(input("How many employees do you want to enter? "))

employees = []

# Input employee details
for i in range(number):
    print("\nEnter Employee", i + 1, "Details:")

    name = input("Enter Employee Name: ")
    id = int(input("Enter Employee ID: "))
    department = input("Enter Employee Department: ")
    salary = int(input("Enter Employee Salary: "))

    # Create object
    emp = Employee(name, id, department, salary)

    # Add object to list
    employees.append(emp)


# Display all employees
print("\n========== ALL EMPLOYEES ==========")

for emp in employees:
    emp.display_info()
```
